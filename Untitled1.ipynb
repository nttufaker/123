```python
import pandas as pd
import matplotlib.pyplot as plt

# 建立人口資料
data = {
    "年份": [2019, 2020, 2021, 2022, 2023, 2024],
    "人口數": [
        23603121,
        23561236,
        23375314,
        23264092,
        23420134,
        23400000
    ]
}

df = pd.DataFrame(data)

# 顯示資料
print(df)

# 計算平均人口
print("平均人口：", df["人口數"].mean())

# 找出人口最多的年份
print("人口最多年份：",
      df.loc[df["人口數"].idxmax(), "年份"])

# 畫折線圖
plt.plot(
    df["年份"],
    df["人口數"],
    marker="o"
)

plt.title("Taiwan Population Trend")
plt.xlabel("Year")
plt.ylabel("Population")
plt.grid()
plt.show()
```
